```python
import ColabTurtle.Turtle as t
from IPython.display import clear_output
import math
import random
```

## Part 1: Basic L-System Implementation

First, let's implement our core L-System functions. These will be used throughout the lab to generate and draw various patterns.

```python
def create_l_system(iterations, axiom, rules):
    """Generate L-System instructions based on axiom and rules."""

    result = axiom

    for _ in range(iterations):

        new_string = ""

        for char in result:
            new_string += rules.get(char, char)

        result = new_string

    return result


def draw_l_system(instructions, angle, distance):
    """Draw L-System instructions using Turtle."""

    stack = []

    for command in instructions:

        if command == "F":

            # Move forward and draw
            t.forward(distance)

        elif command == "f":

            # Move forward without drawing
            t.penup()
            t.forward(distance)
            t.pendown()

        elif command == "+":

            # Turn right
            t.right(angle)

        elif command == "-":

            # Turn left
            t.left(angle)

        elif command == "[":

            # Save current position and direction
            stack.append(
                (t.position(), t.heading())
            )

        elif command == "]":

            # Return to saved position and direction
            position, heading = stack.pop()

            t.penup()
            t.goto(position)
            t.setheading(heading)
            t.pendown()


def setup_turtle():

    clear_output(wait=True)

    t.initializeTurtle(
        initial_speed=13,
        initial_window_size=(800, 800)
    )

    t.hideturtle()
    t.speed(13)

    t.bgcolor("black")
    t.pencolor("white")
    t.width(1)
```

## Pattern 1: Orb Web

This pattern creates a traditional circular spider web. An L-System
rule generates the repeated radial structure, while procedural rules
connect the spokes with layers of capture threads.

Parameters such as the number of spokes, number of rings, radius,
and color can be adjusted to produce different variations.

```python
setup_turtle()


# ADJUSTABLE PARAMETERS

canvas_width = 800
canvas_height = 800

spokes = 14          # Number of radial support threads
rings = 11           # Number of connecting web layers

size = 0.70          # Web size relative to available canvas space

web_color = "white"
line_width = 1


# L-SYSTEM RULE 1

axiom = "X"

rules = {
    "X": "F+X"
}

iterations = spokes

# Divide a full circle evenly between the spokes
angle = 360 / spokes


# Generate the L-System instructions

instructions = create_l_system(
    iterations,
    axiom,
    rules
)

print("Pattern 1: Orb Web")
print("L-System Rule: X -> F+X")
print("L-System instructions:")
print(instructions)


# Find the center of the canvas
center_x = canvas_width / 2
center_y = canvas_height / 2

# Find the shortest distance from the center
# to the edge of the canvas
available_space = min(
    center_x,
    center_y,
    canvas_width - center_x,
    canvas_height - center_y
)

# Calculate a safe web radius
web_radius = available_space * size


t.pencolor(web_color)
t.width(line_width)


# DRAW L-SYSTEM RADIAL SPOKES

t.penup()
t.goto(center_x, center_y)

# Start facing upward
t.setheading(90)

t.pendown()


for command in instructions:

    if command == "F":

        # Draw one spoke outward
        t.forward(web_radius)

        # Return to the center without drawing
        t.penup()
        t.goto(center_x, center_y)
        t.pendown()

    elif command == "+":

        # Rotate before drawing the next spoke
        t.right(angle)


# DRAW CONNECTING WEB THREADS

angle_step = 360 / spokes

# Leave a small open area in the middle
inner_radius = web_radius * 0.10

# Calculate spacing between each web layer
ring_spacing = (
    web_radius - inner_radius
) / rings

for ring in range(rings):

    radius = (
        inner_radius
        + ring * ring_spacing
    )

    # Slightly alternate the layers
    # so the web does not look perfectly mechanical
    offset = (ring % 2) * 2

    for i in range(spokes):

        # Angle of current spoke
        angle1 = i * angle_step

        # Angle of next spoke
        angle2 = (i + 1) * angle_step


        # Start point
        theta1 = math.radians(angle1)
        x1 = (
            center_x
            + radius * math.cos(theta1)
        )
        y1 = (
            center_y
            + radius * math.sin(theta1)
        )

        # End point
        theta2 = math.radians(angle2)
        x2 = (
            center_x
            + radius * math.cos(theta2)
        )
        y2 = (
            center_y
            + radius * math.sin(theta2)
        )

        # Mid point
        middle_angle = math.radians(
            (angle1 + angle2) / 2
        )

        # Pull midpoint slightly toward the center
        middle_radius = (
            radius * 0.96 - offset
        )
        middle_x = (
            center_x
            + middle_radius
            * math.cos(middle_angle)
        )

        middle_y = (
            center_y
            + middle_radius
            * math.sin(middle_angle)
        )

        # Draw connecting thread
        t.penup()
        t.goto(x1, y1)
        t.pendown()

        t.goto(middle_x, middle_y)
        t.goto(x2, y2)
```

## Pattern 2: Corner Web

This pattern creates a web that grows from a corner rather than a
central circular structure. The support threads spread across a
90-degree area and are connected by curved capture threads.

Changing the number of spokes, rings, and web radius changes the
density and size of the web.

```python
setup_turtle()


# ADJUSTABLE PARAMETERS

canvas_width = 800
canvas_height = 800

iterations = 4       # Number of L-System generations
rings = 9            # Number of connecting web layers

size = 0.65          # Web size relative to available canvas space

web_color = "cyan"
line_width = 1


axiom = "X"

rules = {
    "X": "F+FX"
}

# Generate the L-System instructions
instructions = create_l_system(
    iterations,
    axiom,
    rules
)
print("Pattern 2: Corner Web")
print("L-System Rule: X -> F+FX")
print("L-System instructions:")
print(instructions)

# Count the F symbols generated by the L-System
generated_spokes = instructions.count("F")
# Add extra support threads so the web has enough structure to be recognizable
spokes = generated_spokes + 5

print("Number of support threads:", spokes)

# CALCULATE POSITION AND SIZE AUTOMATICALLY
margin = min(
    canvas_width,
    canvas_height
) * 0.10

# Place the starting point based on canvas size
center_x = margin
center_y = margin
available_width = canvas_width - center_x - margin
available_height = canvas_height - center_y - margin
web_radius = min(
    available_width,
    available_height
) * size

t.pencolor(web_color)
t.width(line_width)

# The corner web spreads across 90 degrees
start_angle = 0
end_angle = 90

angle = (
    end_angle - start_angle
) / (spokes - 1)

# DRAW L-SYSTEM GENERATED SUPPORT THREADS
for i in range(spokes):
    current_angle = (
        start_angle
        + i * angle
    )

    # Return to the starting corner
    t.penup()
    t.goto(center_x, center_y)
    # Turn toward the next direction
    t.setheading(current_angle)
    t.pendown()
    # Draw one support thread
    t.forward(web_radius)

# DRAW CONNECTING WEB THREADS
ring_spacing = web_radius / rings

for ring in range(1, rings + 1):
    radius = ring * ring_spacing
    points = []
    for i in range(spokes):

        current_angle = (
            start_angle
            + i * angle
        )

        # Find point on current support thread
        theta = math.radians(
            current_angle
        )
        x = (
            center_x
            + radius * math.cos(theta)
        )
        y = (
            center_y
            + radius * math.sin(theta)
        )
        points.append((x, y))

    # Connect each support thread to the next support thread
    for i in range(len(points) - 1):

        # Start point
        x1, y1 = points[i]

        # End point
        x2, y2 = points[i + 1]

        # Mid point
        middle_x = (x1 + x2) / 2
        middle_y = (y1 + y2) / 2

        # Pull midpoint slightly toward the corner
        middle_x += (
            center_x - middle_x
        ) * 0.04

        middle_y += (
            center_y - middle_y
        ) * 0.04

        # Draw connecting thread
        t.penup()
        t.goto(x1, y1)
        t.pendown()

        t.goto(middle_x, middle_y)
        t.goto(x2, y2)
```

## Pattern 3: Irregular / Damaged Web

This pattern explores a less organized spider web. Random variation
changes the angle and length of the support threads, while some
connecting threads are intentionally skipped.

The random seed keeps the generated result reproducible while still
allowing different seeds to create different web structures.

```python
setup_turtle()

# ADJUSTABLE PARAMETERS
canvas_width = 800
canvas_height = 800
iterations = 3       # Number of L-System generations
rings = 8            # Number of connecting web layers
size = 0.70          # Web size relative to available canvas space
web_color = "lightblue"
line_width = 1
random_seed = 7      # Change this for a different web
angle_variation = 7  # Maximum random angle variation
thread_variation = 5 # Random movement of connecting threads


# L-SYSTEM RULE 3

axiom = "X"

rules = {
    "X": "F[+X][-X]"
}

# Generate the L-System instructions

instructions = create_l_system(
    iterations,
    axiom,
    rules
)
print("Pattern 3: Damaged Web")
print("L-System Rule: X -> F[+X][-X]")
print("L-System instructions:")
print(instructions)

# Count F symbols generated by the L-System
generated_branches = instructions.count("F")
spokes = generated_branches + 4
print("Number of support threads:", spokes)

# Find center of canvas
center_x = canvas_width / 2
center_y = canvas_height / 2

# Find shortest distance from center to any canvas edge
available_space = min(
    center_x,
    center_y,
    canvas_width - center_x,
    canvas_height - center_y
)

# Leave extra room for random movement
safety_margin = (
    thread_variation + 10
)

# Calculate safe web radius
web_radius = (
    available_space
    - safety_margin
) * size

t.pencolor(web_color)
t.width(line_width)

# Change random_seed to create a new variation
random.seed(random_seed)

# Divide the full circle between the spokes
angle = 360 / spokes
spoke_data = []

# DRAW IRREGULAR RADIAL SPOKES
for i in range(spokes):
    # Give each spoke a slightly different angle
    angle_offset = random.uniform(
        -angle_variation,
        angle_variation
    )

    current_angle = (
        i * angle
        + angle_offset
    )

    # Give each spoke a different length
    spoke_length = random.uniform(
        web_radius * 0.75,
        web_radius
    )

    # Save angle and length for later
    spoke_data.append(
        (current_angle, spoke_length)
    )

    # Return to center
    t.penup()
    t.goto(center_x, center_y)

    # Turn toward current direction
    t.setheading(current_angle)
    t.pendown()

    # Draw one support thread
    t.forward(spoke_length)

# DRAW CONNECTING WEB THREADS
for ring in range(1, rings + 1):
    progress = ring / rings
    points = []

    for current_angle, spoke_length in spoke_data:
        radius = (
            spoke_length * progress
        )

        # Find point on current support thread
        theta = math.radians(
            current_angle
        )
        x = (
            center_x
            + radius * math.cos(theta)
        )
        y = (
            center_y
            + radius * math.sin(theta)
        )
        points.append((x, y))

    for i in range(spokes):
        # Randomly skip some threads to create damaged areas
        if random.random() < 0.18:
            continue

        # Start point
        x1, y1 = points[i]

        # End point
        x2, y2 = points[
            (i + 1) % spokes
        ]

        # Mid point with controlled random movement
        middle_x = (
            (x1 + x2) / 2
            + random.uniform(
                -thread_variation,
                thread_variation
            )
        )

        middle_y = (
            (y1 + y2) / 2
            + random.uniform(
                -thread_variation,
                thread_variation
            )
        )

        # Draw connecting thread
        t.penup()
        t.goto(x1, y1)
        t.pendown()

        t.goto(middle_x, middle_y)
        t.goto(x2, y2)
```
